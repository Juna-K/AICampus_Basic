# **[실습 1] 단층 퍼셉트론 만들기**

**Q1. 다음과 같은 데이터를 준비한다.**
- X (입력 데이터) = [10, 4]
- bias (절편) = [1]
- numpy array로 준비한다.

```python
import numpy as np
# 주어진 데이터 (X1=10, X2=4, y=1)
X = np.array([[10, 4]])
bias = np.array([1])
y = np.array([1])
```

**Q2. 단층 퍼셉트론을 구현한다.**

머신 러닝의 모델을 만드는것과 비슷한 구성으로 만들어보자.

**__init__**
- self.w : np.zeros로 0으로 초기화한다. (입력 데이터의 개수와 동일하게 준비한다.)
- self.bias : 0으로 초기화한다.

**step function**
- threshold를 50으로 잡고, theshold 미만이면 0, 아니면 1을 반환한다.

**predict**
- 선형방적식으로 값을 구한다. (wx + b)
- 그리고 해당 값을 step function에 넣은 결과를 반환한다.

**fit**
- epoch번동안 아래 과정을 반복한다.
- X의 데이터를 순서대로 하나씩 넣고, predict를 해서 예측 결과를 확인한다.
- 실제 결과와 예측 결과의 차이를 계산한다.
- 그리고 각 가중치와 절편을 다음과 같이 업데이트한다.
  weight += learning rate * (실제결과와 예측 결과의 차이) * i번째 데이터의 값

각 epoch마다 가중치와 절편이 어떻게 바뀌었는지 출력한다.

```python
class Perceptron:
    def __init__(self, X, bias, y):
        self.X = X
        self.bias_input = bias
        self.y = y

        self.w = np.zeros(X.shape[1])   # array([0., 0.])
        self.bias = 0                   # 학습할 편향 가중치


    def step_function(self, score):
        # np.where(조건, 참일 때 값, 거짓일 때 값)
        # return np.where(x >= 50, 1, 0)

        threshold = 50

        if score < threshold :
            result = 0
        else:
            result = 1

        return result


    def predict(self, input_X):
        # 두 벡터의 내적 - np.dat() : self.w[0] * input_X[0] + self.w[1] * input_X[1] + self.bias
                    # 퍼셉트론에서 각 입력값에 가중치를 곱하고 합산할 때 사용
        score = np.dot(self.w, input_X) + self.bias      # y = mx + b
        return self.step_function(score)


    def fit(self, epochs, learning_rate):
        for epoch in range(epochs):
            for input_X, actual_y in zip(self.X, self.y):      # 각 epoch별 '하나의' 데이터셋을 기반으로 예측 및 업데이트

                predicted_y = self.predict(input_X)           # 1. 현재 가중치와 편향으로 예측

                error = actual_y - predicted_y                 # 2. 실제 정답과 예측값의 차이

                                                            # 4. 손실 구하기 : MSE / CE를 사용하는 시점
                                                            # 4-2. 손실 기반으로 gradient
                
                self.w += learning_rate * error * input_X    # 5. 오차를 기반으로 가중치와 편향 업데이트
                self.bias += learning_rate * error


            print(                                          # 한 epoch의 모든 데이터를 학습한 뒤 출력
                f"epoch {epoch + 1}: "
                f"가중치 = {self.w}, 편향 = {self.bias}" )
```

**Q3. Learning rate = 0.1, epoch = 5로 학습시킨 후, 각 가중치의 값을 찾아라.**
- 제출을 위해 각 가중치의 값을 기록해둔다.

```python
aa = Perceptron(X, bias, y)
aa.fit(epochs=5, learning_rate=0.1)

print("최종 예측:", aa.predict(X[0]))
```

- 제출을 위해 각 가중치의 값을 기록해둔다.

```python
aa = Perceptron(X, bias, y)
aa.fit(epochs=5, learning_rate=0.01)

print("최종 예측:", aa.predict(X[0]))
```

### **1. AND**
단일 퍼셉트론을 활용하여 AND 문제를 해결해보자.

**Q1. AND Gate의 입력과 출력 데이터를 준비한다.**

**Q2. 단층 퍼셉트론을 생성하고, 학습시킨다.**

- epoch : 100
