```python
import numpy as np
import pandas as pd

df = pd.read_csv("input.csv")
ans = df.isnull().sum()

ans
```

```python
ans_df = pd.DataFrame(ans, columns=["null"])
ans_df
```

```python
ans_df.to_csv("output.csv", index=False)

# from google.colab import files
# files.download("output.csv")
```

## 2.

```python
submission = []   # 제출 리스트
```

```python
q1 = df["영어"].mean().round(2)

submission.append(q1)

print(q1, submission)
```

```python
q2 = df["성별"].value_counts(normalize=True).round(2)["남"]

submission.append(q2)

print(q2, submission)
```

```python
# csv를 뽑기 위해 dataframe으로 변환
ans = pd.DataFrame(submission, columns=["ans"])

# ans(데이터프레임)을 output.csv 파일로 저장
ans.to_csv("output.csv", index=False)
```
