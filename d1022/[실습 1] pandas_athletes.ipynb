# **[athletes.csv] 실습 문제**


```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/dukalee/pandas_exercises/refs/heads/main/athletes.csv")
```

```python
df.head()
```

###**Q1. medals가 가장 많은 선수는 누구인가?**

```python
ans = []

# (df['medals'] == df['medals'].max())
# df[(df['medals'] == df['medals'].max())]['name'].values  # ['Daniel Lee']
q1 = df[(df['medals'] == df['medals'].max())]['name'].values[0]  # 'Daniel Lee'
ans.append(q1)

ans
```

### **Q2. country가 'South Korea'인 선수는 몇 명인가?**

```python
# df['country'] == 'South Korea'

# len(df[df['country'] == 'South Korea'])
q2 = (df['country'] == 'South Korea').sum()

ans.append(q2)
```

###**Q3. medals가 0인 선수는 몇 명인가?**

```python
q3 = len(df[df['medals'] == 0])

ans.append(q3)
```

###**Q4. 'South Korea' 선수들의 medals 총합은?**

```python
q4 = df[df['country'] == 'South Korea']['medals'].sum()
ans.append(q4)
```

```python
print(q4)
```

###**Q5. age가 가장 어린 선수는 누구인가?**

```python
q5 = df[df['age'] == df['age'].min()]['name'].values[0]

ans.append(q5)
```

###**Q6. sport가 'Archery'인 선수의 age는?**

```python
q6 = df[df['sport']=='Archery']['age'].values[0]
ans.append(q6)
```

###**Q7. South Korea 선수들의 평균 나이는? (소수점 2자리까지)**

```python
q7 = round(df[df['country']=='South Korea']['age'].mean(),2)
ans.append(q7)
```

###**Q8. medals가 2개 이상인 선수는 몇 명인가?**

```python
q8 = len(df[df['medals']>=2])
ans.append(q8)
```

### **Q9. age 기준 상위 3명 중 medals가 가장 많은 선수는 누구인가?**

```python
q9 = df.sort_values(by='age', ascending=False)[:3]

q9 = q9[q9['medals'] == q9['medals'].max()]['name'].values[0]

ans.append(q9)
```

###**Q10. 'South Korea'가 아닌 선수들의 평균 medals는? (소수 둘째 자리 반올림)**

```python
q10 = round(df[df['country']!='South Korea']['medals'].mean(), 2)
ans.append(q10)
```

## **제출**

```python
df = pd.DataFrame(ans, columns=['ans'])
df.to_csv("output_1.csv", index=False)
```
