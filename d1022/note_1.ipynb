### 원하는 정보만 추출하기 (indexing)

```python
# indexing : loc, iloc, [열] 으로 접근
df[2:4]
```

```python
# 1. fancy indexing: 원하는 것만 딱 pick해서 볼 수 있음
df.loc[[0, 2, 4], ['나이', '급여']]    # [[행], [열]]

# df.loc[[i for i in df.index if i % 2 == 0]] 
      # [표현식 for 변수 in 반복가능한객체 if 조건]
```

```python
# 2. boolean indexing: 조건별 indexing (조건 충족 데이터만 추출)
    # df[조건식]    # True(1)인 데이터만
print(df[df['나이'] < 30], '\n')
print(df[df['몸무게'] <= 80], '\n')
```

```python
# 조건 결합 : and(&) or(|) not(~)
df[(df['나이']<30) & (df['급여']<=4000000)]
```

### DataFrame에서 주로 사용하는 Method

```python
# df.min()
# df.max()
# df.sum()
# df.mean() 
# df.std() : 표준편차
# df.count() : 개수
# df.mode() : 최빈값 - 결측치 처리에 사용

df[(df['급여']) == (df['급여'].max())]
```

```python
# 한 열의 정보 정렬 (* 원본데이터는 수정되지 않음)
df['나이'].value_counts(normalize=True)
df['나이'].unique()
df['나이'].nunique()
df['급여'].sort_values()                    # 오름차순
df['급여'].sort_values(ascending=False)     # 내림차순
```

```python
# 행을 추가
player = {
    '이름' : '홍길동',
    '나이' : 99,
    '몸무게' : 876,
    '급여' : 0,
    '새로운열' : 1,
    '키' : 800
}

p = pd.DataFrame([player])  #하나만 있어도 배열[]로 전달헤애 함
p
```

```python
# pd.concat([원본 데이터프레임, 추가할 데이터프레임])
dc = pd.concat([dc, p])
dc = pd.concat([dc, p], ignore_index=True)
print(dc)
```

```python
new_column = pd.DataFrame({
    "a" : [1, 2, 3,4,5,6,7,8],
    "b" : [10,11,12,13,14,15,16,17]
})
new_column
```

```python
# dataframe을 열쪽에 붙이기 - axis=1
dc = pd.concat([dc, new_column], axis=1)
dc
```

```python
# 행 삭제
dc = dc.drop(7)

# 열 삭제 - axis=1
dc = dc.drop('새로운열', axis=1)

dc
```
